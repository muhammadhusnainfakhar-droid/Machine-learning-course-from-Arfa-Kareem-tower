#### Dictionary  ....stores key , value pairs

```python
mymarks={"Math":90,"Chem":89, "Phy":87}
type(mymarks)
```

```python
mymarks.keys() #return all keys
```

```python
mymarks.values() #return all values
```

```python
mymarks["Chem"]  #to access single value by using keysbas index
```

```python
#loop through a dictionary
for k in mymarks.keys():
    print(f"Marks in {k}:{mymarks[k]}")
```

```python
#find the average of marks
summ=0
for k in mymarks.keys():
    summ=summ+mymarks[k]
avg=summ/len(mymarks)
print(avg)
```

```python
# Add/update new element into dictionary
mymarks["Computer Science"]=98
```
