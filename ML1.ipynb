MATCH CASE STATEMENTS

```python
day=int(input("Enter a day number ."))
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thrusday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid choice")
        
```

```python
num1=int(input("Enter a num."))
num2=int(input("Enter a num"))
operation=input("Enter operation")
match operation:
    case "add":
        print(f"sum of two numbers {num1} and {num2} is {num1 + num2}")
    case "subtract":
        print(f"subtraction of {num2} from {num1} is {num1 - num2}")
    case "multiply":
        print(f"multiplication of two numbers {num1} and {num2} is {num1 * num2}")
    case "divide":
        print(f"division of two numbers {num1} and {num2} is {num1 / num2}")
    case "power":
        print(f"power of {num1} to {num2} is {num1 * num2}")
    case _:
        print("Invalid choice")
         
         
```
