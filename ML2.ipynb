```python
marks=eval(input("Enter your FSC marks"))
total_marks=eval(input("Enter total marks"))
percent=(marks/total_marks)*100
print(f"Your percentage is{percent}")
if percent<=50:
    print("Grade is F")
elif (percent>50) and (percent<=60):
    print("Grade is D")
elif (percent>60) and (percent<=70):
    print("Grade is C")
elif (percent>70) and (percent<=80):
    print("Grade is B")
elif (percent>80) and (percent<=90):
    print("Grade is A")
elif (percent>90) and (percent<=100):
    print("Grade is A+")
else:
    print("Invalid input")
```
